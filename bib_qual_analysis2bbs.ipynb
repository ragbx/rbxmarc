## Etude actions vendangeurs

```python
rbx_bib_qual.groupby(['bib_pat', 'bib_rbx_vdg_action', 'bib_rbx_support'])['nb'].sum()
```

remettre d'équerre comme suit :
- aucune action : tous les documents patrimoniaux et les périodiques, sauf ceux pour lesquels il est déjà spécifié une autre valeur, + matériel, jeu
- autorités seulement : tous les documents non patrimoniaux DVD et disques
- notice bibliographique et autorités : tous les documents non patrimoniaux hors périodiques et audiovisuel

### Hors patrimoine

```python
no_pat_bib_aut = rbx_bib_qual[rbx_bib_qual['bib_pat'] == False]
no_pat_bib_aut = no_pat_bib_aut[~no_pat_bib_aut['bib_rbx_support'].isin(['Périodique', 'Périodique - article',
                                                    'DVD', 'VHS, UMATIC ou film',
                                                    'Cassette audio', 'Disque compact', 'Disque gomme-laque', 'Disque microsillon',
                                                    'Jeu', 'Matériel'
                                                   ])]
no_pat_bib_aut2corr = no_pat_bib_aut[no_pat_bib_aut['bib_rbx_vdg_action'] != 'notice bibliographique et autorités']
biblionumber_bibaut.extend(no_pat_bib_aut2corr['bib_record_id'].to_list())
print(no_pat_bib_aut.groupby(['bib_rbx_support'])['nb'].sum())
print("---------")
no_pat_bib_aut2corr.groupby(['bib_rbx_support'])['nb'].sum()
```

```python
no_pat_aut = rbx_bib_qual[rbx_bib_qual['bib_pat'] == False]
no_pat_aut = no_pat_aut[no_pat_aut['bib_rbx_support'].isin([
                                                    'CD-ROM',
                                                    'DVD', 'VHS, UMATIC ou film',
                                                    'Cassette audio', 'Disque compact', 'Disque gomme-laque', 'Disque microsillon',
                                                   ])]
no_pat_aut2corr = no_pat_aut[no_pat_aut['bib_rbx_vdg_action'] != 'autorités seulement']
biblionumber_aut.extend(no_pat_aut2corr['bib_record_id'].to_list())
print(no_pat_aut.groupby(['bib_rbx_support'])['nb'].sum())
print("---------")
no_pat_aut2corr.groupby(['bib_rbx_support'])['nb'].sum()
```

```python
no_pat_no_vdg = rbx_bib_qual[rbx_bib_qual['bib_pat'] == False]
no_pat_no_vdg = no_pat_no_vdg[no_pat_no_vdg['bib_rbx_support'].isin([
                                                    'Jeu', 'Matériel',
                                                    'Périodique', 'Périodique - article'
                                                   ])]
no_pat_no_vdg2corr = no_pat_no_vdg[no_pat_no_vdg['bib_rbx_vdg_action'] != 'aucune action']
print(no_pat_no_vdg.groupby(['bib_rbx_support'])['nb'].sum())
print("---------")
no_pat_no_vdg2corr.groupby(['bib_rbx_support'])['nb'].sum()
```

### Patrimoine

```python
pat_bib_aut = rbx_bib_qual[rbx_bib_qual['bib_pat'] == True]
pat_bib_aut = pat_bib_aut[pat_bib_aut['bib_rbx_vdg_action'] == 'notice bibliographique et autorités']
pat_bib_aut = pat_bib_aut[pat_bib_aut['bib_rbx_support'].isin(['Livre'])]
pat_bib_aut2corr = pat_bib_aut[pat_bib_aut['bib_rbx_vdg_action'] != 'notice bibliographique et autorités']
print(pat_bib_aut.groupby(['bib_rbx_support'])['nb'].sum())
print("---------")
pat_bib_aut2corr.groupby(['bib_rbx_support'])['nb'].sum()
```

```python
pat_aut = rbx_bib_qual[rbx_bib_qual['bib_pat'] == True]
#pat_aut = pat_aut[pat_aut['bib_rbx_vdg_action'] == 'notice bibliographique et autorités']

pat_aut = pat_aut[pat_aut['bib_rbx_support'].isin(['CD-ROM',
                                                   'DVD', 'VHS, UMATIC ou film',
                                                   'Cassette audio', 'Disque compact', 'Disque gomme-laque', 'Disque microsillon'])]
pat_aut2corr = pat_aut[pat_aut['bib_rbx_vdg_action'] != 'autorités seulement']
print(pat_aut.groupby(['bib_rbx_support'])['nb'].sum())
print("---------")
pat_aut2corr.groupby(['bib_rbx_support'])['nb'].sum()
```

### Synthèse

```python
biblionumber_novdg = []
#biblionumber_novdg.extend(pat_no_vdg2corr['bib_record_id'].to_list())
biblionumber_novdg.extend(no_pat_no_vdg2corr['bib_record_id'].to_list())
print("Corrections -- aucune action")
print(len(biblionumber_novdg))
print("------------")

biblionumber_aut = []
biblionumber_aut.extend(pat_aut2corr['bib_record_id'].to_list())
biblionumber_aut.extend(no_pat_aut2corr['bib_record_id'].to_list())
print("Corrections -- autorités seulement")
print(len(biblionumber_aut))
print("------------")

biblionumber_bibaut = []
biblionumber_bibaut.extend(pat_bib_aut2corr['bib_record_id'].to_list())
biblionumber_bibaut.extend(no_pat_bib_aut2corr['bib_record_id'].to_list())
print("Corrections -- notice bibliographique et autorités")
print(len(biblionumber_bibaut))
```

```python
vdg_corr = rbx_bib_qual.copy()

vdg_corr.loc[vdg_corr['bib_record_id'].isin(biblionumber_novdg), 'bib_rbx_vdg_action'] = 'aucune action'
vdg_corr.loc[vdg_corr['bib_record_id'].isin(biblionumber_aut), 'bib_rbx_vdg_action'] = 'autorités seulement'
vdg_corr.loc[vdg_corr['bib_record_id'].isin(biblionumber_bibaut), 'bib_rbx_vdg_action'] = 'notice bibliographique et autorités'

vdg_corr.groupby(['bib_pat', 'bib_rbx_vdg_action', 'bib_rbx_support'])['nb'].sum()
```

```python
pd.DataFrame({'biblionumber' : biblionumber_novdg}).to_csv("extractions/correction_vdg_novdg.csv", header=False, index=False)
pd.DataFrame({'biblionumber' : biblionumber_aut}).to_csv("extractions/correction_vdg_aut.csv", header=False, index=False)
pd.DataFrame({'biblionumber' : biblionumber_bibaut}).to_csv("extractions/correction_vdg_bibaut.csv", header=False, index=False)
```
