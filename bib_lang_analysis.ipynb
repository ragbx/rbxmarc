```python
import pandas as pd
from os.path import join
```

```python
date_export = "2024-08-25"
date_export2 = date_export.replace("-", "")
data_file = join("extractions", f"bib_lang_{date_export2}.csv.gz")


rbx_bib_lang = pd.read_csv(data_file)
rbx_bib_lang['nb'] = 1
```

```python
rbx_bib_lang.columns
```

```python
rbx_bib_lang.groupby(['bib_langue_document'])['nb'].sum().reset_index().sort_values(by='nb', ascending=False)
```

```python
rbx_bib_lang_dut = rbx_bib_lang[rbx_bib_lang['bib_langue_document'].isin(['dut', 'dum'])]
rbx_bib_lang_dut_pat = rbx_bib_lang_dut[rbx_bib_lang_dut['bib_pat'] == True]
rbx_bib_lang_dut_pat
```
