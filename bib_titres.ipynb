```python
import pandas as pd
```

```python
df = pd.read_csv("extractions/extract_titres_20250302.csv.gz")
```

```python
df.loc[(~df['bib_ark_bnf'].isna()) & (df['bib_ark_bnf'].str.contains('ark')), 'bib_alignement_bnf'] = True
```

```python
bdseries_j = pd.read_csv("extractions/biblionumber_jseries.csv")
bdseries_a = pd.read_csv("extractions/biblionumber_aseries.csv")
mangas_j = pd.read_csv("extractions/biblionumber_jmangas.csv")
mangas_a = pd.read_csv("extractions/biblionumber_amangas.csv")

df.loc[df['bib_record_id'].astype(str).isin(mangas_j['biblionumber'].astype(str)), 'public'] = 'jeunesse'
df.loc[df['bib_record_id'].astype(str).isin(mangas_a['biblionumber'].astype(str)), 'public'] = 'adultes'
df.loc[df['bib_record_id'].astype(str).isin(bdseries_j['biblionumber'].astype(str)), 'public'] = 'jeunesse'
df.loc[df['bib_record_id'].astype(str).isin(bdseries_a['biblionumber'].astype(str)), 'public'] = 'adultes'

df.loc[df['bib_record_id'].astype(str).isin(mangas_j['biblionumber'].astype(str)), 'type'] = 'mangas'
df.loc[df['bib_record_id'].astype(str).isin(mangas_a['biblionumber'].astype(str)), 'type'] = 'mangas'
df.loc[df['bib_record_id'].astype(str).isin(bdseries_j['biblionumber'].astype(str)), 'type'] = 'bd_serie'
df.loc[df['bib_record_id'].astype(str).isin(bdseries_a['biblionumber'].astype(str)), 'type'] = 'bd_serie'
```

```python
for c in ['B200ae', 'B200h', 'B200i', 'B200i', 'B225a', 'B225v', 'B225x',
          'B410t', 'B410v', 'B461t', 'B461v', 'B500a', 'B503a', 'B510a', 'B515a', 'B517a']:
    c2 = c + "_"
    df[c2] = False
    df.loc[~df[c].isna(), c2] = True
df.columns
```

```python
df.groupby(['B200ae_', 'B200h_',
       'B200i_', 'B225a_', 'B225v_', 'B225x_', 'B410t_', 'B410v_', 'B461t_',
       'B461v_', 'B500a_', 'B503a_', 'B510a_', 'B515a_', 'B517a_'])['bib_record_id'].count().to_csv("resultats/titres_stat.csv")
```

```python
bd = df[~df['type'].isna()]
bd.groupby(['bib_agence_cat', 'B200ae_', 'B200h_', 'B200i_', 'B410t_',
       'B410v_', 'B461t_', 'B461v_'])['bib_record_id'].count().to_csv("resultats/titres_bd.csv")
```
