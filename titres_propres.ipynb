```python
from os.path import join
import pandas as pd

from pymarc import MARCReader
from rbxmarc import Rbxmrc, Rbxbib2dict
from os.path import join


rbxmrc = Rbxmrc()
referentiels = rbxmrc.referentiels

marc_file = join("data", "2025-03-09-notices_total.mrc")

with open(marc_file, 'rb') as fh:
    metadatas = []
    reader = MARCReader(fh, to_unicode=True, force_utf8=True)
    i = 0
    for record in reader:
        met = {}
        fields = record.get_fields("001")
        for field in fields:
            met['biblionumber'] = field.data

        f200a_value = []
        fields = record.get_fields("200")
        for field in fields:
            if hasattr(field, "subfields"):
                for subfield in field.subfields:
                    if subfield.code == "a":
                        f200a_value.append(subfield.value)
        met['nb_titres'] = len(f200a_value)
        print(met)
        metadatas.append(met)
```

```python
df = pd.DataFrame(metadatas)
```

```python
df = df.sort_values(by='nb_titres', ascending=False)
```
